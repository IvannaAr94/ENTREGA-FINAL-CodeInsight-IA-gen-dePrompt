## 5.1 Técnicas de Fast Prompting utilizadas

Durante el desarrollo de CodeInsight AI se aplicarán diferentes técnicas de Fast Prompting para mejorar la calidad, precisión y estructura de las respuestas generadas por el modelo.

### Role Prompting

Se asignará al modelo un rol específico, por ejemplo, el de desarrollador web senior o especialista en seguridad. Esto permite orientar la respuesta hacia un contexto técnico determinado.

**Ejemplo:**

> Actúa como un desarrollador web senior especializado en revisión de código y buenas prácticas.

---

### Context Prompting

Se proporcionará información sobre el proyecto analizado para que el modelo pueda interpretar correctamente el código y generar recomendaciones más relevantes.

**Ejemplo:**

> El código pertenece a una aplicación web de comercio electrónico desarrollada como proyecto académico.

---

### Task Specification

Se indicará claramente cuál es la tarea que debe realizar el modelo, evitando instrucciones ambiguas.

**Ejemplo:**

> Analiza el siguiente fragmento de código e identifica problemas relacionados con mantenibilidad, seguridad y rendimiento.

---

### Constraints

Se establecerán límites y reglas que deberá respetar la respuesta.

**Ejemplo:**

> Devuelve como máximo cinco hallazgos y no inventes vulnerabilidades que no puedan justificarse a partir del código proporcionado.

---

### Output Formatting

Se definirá previamente la estructura esperada de la respuesta para facilitar su interpretación y posterior procesamiento.

**Ejemplo de formato:**

- Hallazgo
- Categoría
- Prioridad
- Explicación
- Recomendación

---

### Few-Shot Prompting

Se proporcionará al modelo uno o más ejemplos del tipo de respuesta esperada antes de solicitar el análisis definitivo.

Esto permite orientar mejor la estructura y el nivel de detalle de la salida.

---

### Prompt Chaining

Una tarea compleja será dividida en diferentes prompts relacionados entre sí.

Los resultados obtenidos en una etapa podrán utilizarse como contexto en las etapas siguientes.

El flujo general será:

**Análisis técnico → Análisis de seguridad → Reporte → Backlog de mejoras**

---

### Optimización de prompts

Se comparará un prompt básico con un prompt optimizado para evaluar cómo la incorporación de contexto, rol, restricciones y formato de salida modifica la utilidad de la respuesta generada.

## 5.2 Justificación de las técnicas seleccionadas

Las técnicas seleccionadas permiten construir prompts más claros y específicos sin necesidad de modificar o entrenar el modelo utilizado.

El **Role Prompting** y el **Context Prompting** ayudan a orientar al modelo hacia el dominio del problema. La especificación de tareas y restricciones reduce respuestas demasiado generales o alejadas del objetivo.

El uso de un formato estructurado facilita la lectura de los resultados y permite reutilizarlos posteriormente para generar reportes, backlog o visualizaciones.

El **Few-Shot Prompting** permite mostrar al modelo ejemplos del resultado esperado, mientras que el **Prompt Chaining** ayuda a dividir procesos complejos en etapas más simples y controlables.

Estas técnicas fueron seleccionadas porque pueden implementarse directamente mediante prompts, requieren pocos recursos adicionales y son adecuadas para desarrollar una Proof of Concept dentro del tiempo y los recursos disponibles.

## 6.3 Función reutilizable para consultas Texto → Texto

Para evitar repetir el código de conexión en cada etapa, se implementa una función reutilizable encargada de enviar prompts al modelo y devolver únicamente el contenido de la respuesta.

Esta función será utilizada posteriormente para el análisis técnico, análisis de seguridad, generación de reportes y creación del backlog de mejoras.

```python
# Función reutilizable para consultar el modelo

def consultar_ia(prompt, temperature=0.2):
    """
    Envía un prompt al modelo configurado y devuelve su respuesta.
    """

    try:
        respuesta = client.chat.completions.create(
            model=MODEL,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=temperature
        )

        return respuesta.choices[0].message.content

    except Exception as error:
        return f"❌ Error al consultar el modelo: {error}"


print("✅ Función consultar_ia() creada correctamente.")
```

- Output Formatting

```python
prompt_optimizado = f"""
ROL:
Actúa como un desarrollador web senior especializado en React,
revisión de código y buenas prácticas de desarrollo frontend.

CONTEXTO:
El siguiente código pertenece a una aplicación web de comercio electrónico.
El componente muestra una lista de productos y permite eliminar un producto
mediante una petición HTTP DELETE.

TAREA:
Analiza exclusivamente el código proporcionado e identifica los principales
problemas técnicos relacionados con:

- calidad del código
- mantenibilidad
- manejo de errores
- experiencia de usuario
- buenas prácticas de React

RESTRICCIONES:
- Devuelve como máximo 5 hallazgos.
- No inventes vulnerabilidades ni problemas que no puedan justificarse
  directamente a partir del código.
- No asumas características del backend que no aparecen en el fragmento.
- Evita recomendaciones genéricas.
- Prioriza los problemas más relevantes.
- Clasifica cada problema como prioridad Alta, Media o Baja.

FORMATO DE SALIDA:

Hallazgo:
Categoría:
Prioridad:
Problema:
Recomendación:

EJEMPLO DEL FORMATO ESPERADO:

Hallazgo: Uso de un identificador inestable
Categoría: React
Prioridad: Media
Problema: El componente utiliza un valor que puede cambiar entre renders.
Recomendación: Utilizar un identificador único y estable asociado al elemento.

CÓDIGO A ANALIZAR:

{codigo_prueba}
"""

print(prompt_optimizado)
```

```python
respuesta_optimizada = consultar_ia(prompt_optimizado)

print("RESPUESTA DEL PROMPT OPTIMIZADO")
print("=" * 60)
print(respuesta_optimizada)
```

### Técnicas aplicadas en el prompt optimizado

El prompt optimizado incorpora diferentes técnicas de Fast Prompting:

- **Role Prompting:** se asigna al modelo el rol de desarrollador web senior especializado en React y revisión de código.
- **Context Prompting:** se explica que el código pertenece a una aplicación de comercio electrónico y cuál es la función del componente.
- **Task Specification:** se especifican claramente las categorías que deben analizarse.
- **Constraints:** se limita la cantidad de hallazgos y se prohíbe inventar problemas no justificables mediante el código.
- **Output Formatting:** se establece una estructura uniforme para todos los hallazgos.
- **Few-Shot Prompting:** se proporciona un ejemplo del formato de respuesta esperado.

Estas técnicas buscan reducir la ambigüedad del prompt y obtener resultados más consistentes, estructurados y reutilizables.

## 6.8 Prompt Chaining

Para resolver tareas más complejas, CodeInsight AI utiliza la técnica de **Prompt Chaining**.

En lugar de solicitar al modelo un análisis completo dentro de un único prompt, el proceso se divide en diferentes etapas especializadas. La salida obtenida en una etapa se utiliza como contexto para la siguiente.

El flujo implementado será:

**Análisis técnico → Análisis de seguridad → Reporte técnico → Backlog de mejoras**

Este enfoque permite organizar mejor las tareas, reducir la complejidad de cada prompt y reutilizar la información generada anteriormente.

### Etapa 1 — Análisis de seguridad

En esta etapa se realiza una revisión orientada específicamente a seguridad.

El modelo deberá analizar solamente los riesgos que puedan justificarse a partir del código proporcionado, evitando inventar vulnerabilidades o asumir configuraciones que no están disponibles.

```python
# Prompt especializado en seguridad

prompt_seguridad = f"""
ROL:
Actúa como un especialista en seguridad de aplicaciones web
con experiencia en revisión de código frontend.

CONTEXTO:
Se está analizando un componente React perteneciente a una
aplicación web de comercio electrónico.

El análisis técnico previo fue:

{respuesta_optimizada}

CÓDIGO ORIGINAL:

{codigo_prueba}

TAREA:
Realiza una revisión específica de seguridad sobre el código.

RESTRICCIONES:
- Analiza únicamente riesgos que puedan justificarse mediante el código.
- No inventes vulnerabilidades.
- No asumas configuraciones del backend.
- No asumas que existe o no autenticación.
- Devuelve como máximo 3 hallazgos.
- Si no existe evidencia suficiente para afirmar un riesgo, indícalo.
- Clasifica cada hallazgo como Alta, Media o Baja.

FORMATO:

Hallazgo:
Prioridad:
Evidencia en el código:
Explicación:
Recomendación:
"""
```

```python
respuesta_seguridad = consultar_ia(prompt_seguridad)

print("ANÁLISIS DE SEGURIDAD")
print("=" * 60)
print(respuesta_seguridad)
```

### Etapa 2 — Generación del reporte técnico

Los resultados del análisis general y del análisis de seguridad se utilizan como contexto para generar un reporte resumido.

El objetivo es transformar los diferentes hallazgos en información más clara y útil para un desarrollador.

```python
# Prompt para generar un reporte técnico consolidado

prompt_reporte = f"""
ROL:
Actúa como líder técnico de un equipo de desarrollo web.

TAREA:
Genera un reporte técnico breve a partir de los dos análisis
realizados previamente.

ANÁLISIS TÉCNICO:

{respuesta_optimizada}

ANÁLISIS DE SEGURIDAD:

{respuesta_seguridad}

INSTRUCCIONES:
- Resume únicamente la información proporcionada.
- No agregues problemas nuevos.
- Identifica los hallazgos más importantes.
- Explica brevemente su impacto.
- Incluye recomendaciones concretas.
- Mantén un lenguaje técnico pero claro.

ESTRUCTURA DEL REPORTE:

1. Resumen general
2. Principales hallazgos
3. Aspectos de seguridad
4. Recomendaciones prioritarias
"""
```

```python
respuesta_reporte = consultar_ia(prompt_reporte)

print("REPORTE TÉCNICO")
print("=" * 60)
print(respuesta_reporte)
```

### Etapa 3 — Generación del backlog de mejoras

Finalmente, el reporte técnico será transformado en un conjunto de tareas concretas.

Esta etapa permite convertir los resultados producidos por la Inteligencia Artificial en acciones que podrían incorporarse a un backlog de desarrollo.

```python
# Prompt para generar un backlog a partir del reporte

prompt_backlog = f"""
ROL:
Actúa como líder técnico encargado de organizar un backlog
de mejoras para un proyecto web.

REPORTE TÉCNICO:

{respuesta_reporte}

TAREA:
Convierte las recomendaciones del reporte en tareas concretas
para un equipo de desarrollo.

RESTRICCIONES:
- No inventes tareas que no estén relacionadas con el reporte.
- Devuelve como máximo 6 tareas.
- Cada tarea debe ser clara y accionable.
- Clasifica la prioridad como Alta, Media o Baja.

FORMATO:

Tarea:
Prioridad:
Objetivo:
Acción recomendada:
"""
```

```python
respuesta_backlog = consultar_ia(prompt_backlog)

print("BACKLOG DE MEJORAS")
print("=" * 60)
print(respuesta_backlog)
```

## 6.9 Visualización de resultados

Para complementar los resultados generados mediante Inteligencia Artificial, se utilizarán Pandas y Matplotlib para organizar y representar visualmente la información obtenida.

En esta etapa se analizarán las prioridades asignadas por el modelo en las respuestas estructuradas del análisis técnico, análisis de seguridad y backlog de mejoras.

La visualización permite obtener una lectura rápida de la distribución de hallazgos según su prioridad: Alta, Media o Baja.

```python
import re

# Función para extraer prioridades de las respuestas generadas por la IA

def extraer_prioridades(texto):
    patron = r"(?:\*\*)?Prioridad(?:\*\*)?\s*:\s*(?:\*\*)?(Alta|Media|Baja)(?:\*\*)?"

    prioridades = re.findall(
        patron,
        texto,
        flags=re.IGNORECASE
    )

    return [prioridad.capitalize() for prioridad in prioridades]


prioridades_tecnico = extraer_prioridades(respuesta_optimizada)
prioridades_seguridad = extraer_prioridades(respuesta_seguridad)
prioridades_backlog = extraer_prioridades(respuesta_backlog)

print("Análisis técnico:", prioridades_tecnico)
print("Análisis de seguridad:", prioridades_seguridad)
print("Backlog:", prioridades_backlog)
```

```python
# Organización de los resultados en un DataFrame

datos_prioridades = {
    "Etapa": [
        "Análisis técnico",
        "Análisis de seguridad",
        "Backlog de mejoras"
    ],
    "Alta": [
        prioridades_tecnico.count("Alta"),
        prioridades_seguridad.count("Alta"),
        prioridades_backlog.count("Alta")
    ],
    "Media": [
        prioridades_tecnico.count("Media"),
        prioridades_seguridad.count("Media"),
        prioridades_backlog.count("Media")
    ],
    "Baja": [
        prioridades_tecnico.count("Baja"),
        prioridades_seguridad.count("Baja"),
        prioridades_backlog.count("Baja")
    ]
}

df_prioridades = pd.DataFrame(datos_prioridades)

df_prioridades
```

```python
# Total de elementos según prioridad

totales_prioridad = df_prioridades[
    ["Alta", "Media", "Baja"]
].sum()

print("Total por prioridad:")
print(totales_prioridad)
```

```python
# Gráfico de distribución general de prioridades

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

totales_prioridad.plot(
    kind="bar"
)

plt.title("Distribución de prioridades - CodeInsight AI")
plt.xlabel("Prioridad")
plt.ylabel("Cantidad de hallazgos y tareas")

plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
# Comparación de prioridades por etapa del análisis

df_grafico = df_prioridades.set_index("Etapa")

df_grafico.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Prioridades detectadas por etapa")
plt.xlabel("Etapa de CodeInsight AI")
plt.ylabel("Cantidad")

plt.xticks(rotation=15)
plt.tight_layout()

plt.show()
```

### Interpretación de la visualización

Las visualizaciones permiten observar cómo se distribuyen las prioridades asignadas por el modelo durante las distintas etapas de CodeInsight AI.

El análisis técnico identifica problemas generales relacionados con el código, mientras que la etapa de seguridad se concentra únicamente en riesgos que pueden justificarse mediante la evidencia disponible.

Posteriormente, el backlog transforma los hallazgos y recomendaciones en acciones concretas que pueden ser priorizadas por un equipo de desarrollo.

Los gráficos no reemplazan el análisis textual realizado por el modelo, sino que permiten resumir visualmente parte de los resultados y facilitar su interpretación.
